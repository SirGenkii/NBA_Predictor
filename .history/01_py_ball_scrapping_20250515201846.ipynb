```python


nba = Client()

# Extraction des équipes NBA
teams = nba.team.TeamList(league_id='00')  # '00' = NBA (vs G-League, WNBA, etc.)

# Affichage du nombre d'équipes
print(f"{len(teams)} équipes récupérées")
# Aperçu des données
print(teams[0])
```

```python
# On met dans un DataFrame pour faciliter l’analyse et le stockage
teams_df = pd.DataFrame(teams)

# Optionnel : on garde les colonnes principales
main_cols = ['team_id', 'full_name', 'abbreviation', 'city', 'state', 'year_founded']
teams_df = teams_df[main_cols]

# Sauvegarde
teams_df.to_csv('nba_teams.csv', index=False)
```

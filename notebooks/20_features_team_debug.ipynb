```python
import os
from pathlib import Path

import polars as pl

from nba_predictor import settings

TEAM_ID = int(os.environ.get("NBA_DEBUG_TEAM_ID", 1610612747))
WINDOW = int(os.environ.get("NBA_DEBUG_WINDOW", 10))

FORM_PATHS = list(Path(settings.data_paths.silver_team_form_windowed).rglob("*.parquet"))
len(FORM_PATHS)
```

```python
df = pl.concat([pl.scan_parquet(str(path)) for path in FORM_PATHS]).filter(pl.col("team_id") == TEAM_ID).collect()
df.tail(5)
```

```python
metrics = [c for c in df.columns if c.endswith(f"_last{WINDOW}")]
df.select(["game_date"] + metrics[:10]).tail(10)
```

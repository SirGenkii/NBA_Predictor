```python
import os
from pathlib import Path
import polars as pl

from nba_predictor import settings

TEAM_ID = int(os.environ.get("NBA_DEBUG_TEAM_ID", 1610612747))
OPP_ID = int(os.environ.get("NBA_DEBUG_OPP_ID", 1610612738))

H2H_PATHS = list(Path(settings.data_paths.silver_matchups_h2h_features).rglob("*.parquet"))
len(H2H_PATHS)
```

```python
df_h2h = (
    pl.concat([pl.scan_parquet(str(path)) for path in H2H_PATHS])
    .filter((pl.col("team_id") == TEAM_ID) & (pl.col("opponent_team_id") == OPP_ID))
    .sort("game_date")
    .collect()
)
df_h2h.tail(5)
```

```python
windows = [c for c in df_h2h.columns if c.startswith("h2h_win_rate_last")]
df_h2h.select(["game_date", "is_win"] + windows).tail(10)
```
